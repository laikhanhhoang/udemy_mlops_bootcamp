## Import data

```python
data = pd.read_csv("../experiments/data/test.csv")
input = data.drop(columns=["quality"])
output = data["quality"]
```

## Supported function

```python
def eval_metrics(actual, pred):
    rmse = np.sqrt(mean_squared_error(actual, pred))
    mae = mean_absolute_error(actual, pred)
    r2 = r2_score(actual, pred)
    return rmse, mae, r2
```

## Load model from run

```python
(rmse, mae, r2) = eval_metrics(output, predictions)
print("  RMSE: %s" % rmse)
print("  MAE: %s" % mae)
print("  R2: %s" % r2)
```
